```python
import usu_routes
from shapely.plotting import plot_line
import matplotlib.pyplot as plt

segments = usu_routes.import_route(usu_routes.GREEN_ROUTE)
colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red', 'tab:purple']
for i, segment_line_string in enumerate(segments):
    color = colors[i % len(colors)]
    plot_line(segment_line_string, add_points=False, color=color)
plt.show()
```
